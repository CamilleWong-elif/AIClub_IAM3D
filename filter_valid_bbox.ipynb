```python
def filter_label(label_path,split_dict,split):
    split_path = os.path.join(label_path,split)
    split_files = os.listdir(split_path)
    print(f"{split.upper()} SPLIT:")
    print(f'Total file pre-processing {len(split_files)}')
    for i in range(len(split_files)):
        file_path = os.path.join(split_path,split_files[i])
        with open(file_path,'r') as f:
            lines = f.readlines()
        for line in lines:
            if len(line.split()) != 5:
                split_dict[split].append(i)
                break 
    print(f'Total files post-processing {len(split_files) - len(split_dict[split])}')

def filter_imgs(split_dict,split):
    pass
                    
```
